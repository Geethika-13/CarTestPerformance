```python
import pandas as pd
df = pd.read_csv('car_price_performance.csv')
df.head()

```

```python
avg_price = df.groupby('Brand')['Price'].mean().sort_values().index
avg_price
```

```python
brand_mapping = {brand: i for i, brand in enumerate(avg_price)}
df['Encoded Brand']=df['Brand'].map(brand_mapping)
# label encoding with average
df.head()
```

```python
 import pandas as pd
 for column in ['Encoded Brand']:
    min_val = df[column].min()
    max_val = df[column].max()
    df['Scaled ' + column] = (df[column] - min_val) / (max_val - min_val)

df.head()
```

```python
import pandas as pd

for column in ['Horsepower', 'Fuel Efficiency']:

    min_val = df[column].min()
    max_val = df[column].max()

    df['Scaled ' + column] = (df[column] - min_val) / (max_val - min_val)

df.head()
```

```python
df['Performance Index'] = df['Horsepower'] * 0.7 + df['Fuel Efficiency'] * 0.3 / df['Price']
df.head()
```

```python
df.to_csv('car_price_performance.csv', index=False)
df.head()
```
